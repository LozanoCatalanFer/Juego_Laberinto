```python
class ElementoMapa:
    """Clase base para los elementos del mapa del laberinto."""
    def __init__(self,padre):
        self.padre=padre
    def entrar(self,alguien):
        pass
class Contenedor(ElementoMapa):
    """Representa una pared dentro del laberinto."""
    def __init__(self):
        self.hijo=[]
        self.orientaciones=[]

    def agregarHijo(self,unEM):
        unEM.padre=self
        self.hijos.append(unEM)

    def agregarOrientacion(self,unaOr):
        self.orientaciones.append(unaOr)

    def obtenerElementoOr(self,unaOr):
        return unaOr.obtenerElementoOrEn(self)
    
    def ponerEnOr(self,unaOr,unEM):
        unaOr.ponerElemento(unEM,self)



    def __str__(self):
        return "Pared"



class Pared(ElementoMapa):
    """Representa una pared dentro del laberinto."""
    def __str__(self):
        return "Pared"

class Puerta(ElementoMapa):
    """Representa una puerta dentro del laberinto."""
    def __init__(self, lado1, lado2, abierta=False):
        self.lado1 = lado1
        self.lado2 = lado2
        self.abierta = abierta
    
    def abrir(self):
        self.abierta = True
    
    def cerrar(self):
        self.abierta = False
    
    def __str__(self):
        return "Puerta Abierta" if self.abierta else "Puerta Cerrada"

class Habitacion(ElementoMapa):
    """Representa una habitación dentro del laberinto."""
    def __init__(self, num):
        self.num = num
        self.norte = None
        self.sur = None
        self.este = None
        self.oeste = None
    
    def establecer_lado(self, direccion, elemento):
        if direccion == "norte":
            self.norte = elemento
        elif direccion == "sur":
            self.sur = elemento
        elif direccion == "este":
            self.este = elemento
        elif direccion == "oeste":
            self.oeste = elemento
    
    def __str__(self):
        return f"Habitación {self.num}"

class Laberinto(ElementoMapa):
    """Representa el laberinto que contiene habitaciones."""
    def __init__(self):
        self.habitaciones = {}
    
    def agregar_habitacion(self, habitacion):
        self.habitaciones[habitacion.num] = habitacion
    
    def obtener_habitacion(self, num):
        return self.habitaciones.get(num, None)
    
    def __str__(self):
        return f"Laberinto con {len(self.habitaciones)} habitaciones"

class Juego:
    """Clase principal que maneja el juego del laberinto."""
    def __init__(self):
        self.laberinto = Laberinto()
        self.bichos=[]
    
    def configurar_laberinto2H(self):
        h1 = Habitacion(1)
        h2 = Habitacion(2)
        p = Puerta(h1, h2)
        
        h1.establecer_lado("este", p)
        h1.establecer_lado("oeste", Pared())
        h1.establecer_lado("norte", Pared())
        h1.establecer_lado("sur", Pared())
        
        h2.establecer_lado("oeste", p)
        h2.establecer_lado("este", Pared())
        h2.establecer_lado("norte", Pared())
        h2.establecer_lado("sur", Pared())
        
        self.laberinto.agregar_habitacion(h1)
        self.laberinto.agregar_habitacion(h2)
    
    def agregarBichos(self,bicho):
        self.bichos.append(bicho)

    def eliminarBichos(self,bicho):
        if bicho in self.bichos:
            del self.bichos[bicho]
        else:
            print("No podemos borrar un bicho que no existe")
    def obtenerHabitacion(self,unNum):
        return self.laberinto.obtener_habitacion(unNum)
    
    def jugar(self):
        print("Bienvenido al juego del Laberinto!")
        print(self.laberinto)

# Ejemplo de uso
if __name__ == "__main__":
    juego = Juego()
    juego.configurar_laberinto()
    juego.jugar()


class Bicho:
    def __init__(self, vida, poder, posicion, modo):
        self.vida = vida
        self.poder = poder
        self.posicion = posicion
        self.modo = modo
    def __str__(self):
        return f"Bicho {self.modo} en {self.posicion} con {self.vida} de vida y {self.poder} de poder"

class BichoPerezoso(Bicho):
    def __init__(self, vida, poder, posicion):
        super().__init__(vida, poder, posicion, "Perezoso")

class BichoAgresivo(Bicho):
    def __init__(self, vida, poder, posicion):
        super().__init__(vida, poder, posicion, "Agresivo")
```
